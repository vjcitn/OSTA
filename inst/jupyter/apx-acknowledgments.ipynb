- This book is built using [**BiocBook**](https://bioconductor.org/packages/release/BiocBook), developed by [Jacques Serizay](https://jserizay.com/) for building Quarto books with Bioconductor. More details on BiocBook can be found in this [demo](https://bioconductor.org/books/release/BiocBookDemo/).

- This book is inspired by the **Orchestrating Single-Cell Analysis with Bioconductor (OSCA)** book [@Amezquita2020-OSCA] on analysis workflows for single-cell data using R and Bioconductor, available [here](https://bioconductor.org/books/release/OSCA/).

- Details on data preprocessing procedures for **10x Genomics Visium** data can be found in our related online book [Visium data preprocessing](https://lmweber.org/Visium-data-preprocessing/).

- **Feedback and suggestions** may be provided as [GitHub](https://github.com/lmweber/OSTA) issues for discussion with the maintainers.
